```python
datos_agrupados = data.groupby(["StoreID", "Producto"], observed=True)["PrecioTotal"].sum().reset_index(), data.groupby()
```

```python
datos_agrupados.to_excel("facturacion.xlsx")
```

```python
import plotly.express as px
```

```python
columna = ["Dia", "Manager", "Locacion", "Producto"]
for columna in columna:
    grafico = px.histogram(data, x = columna, 
                y = "PrecioTotal",
                title = "Facturacion",
                color = "StoreID")
    grafico.show()
    grafico.write_html(f"facturacion_por {columna}.html")
```

```python
fig = px.bar(
    x = "Location",
    y = "PrecioTotal",
    color = "StoreID",
    text_auto = True,
    range_x = [0,1000000],
    animation_frame = "Fecha"
)
fig.show()
fig.write_html("grafico_animado.html")
```
