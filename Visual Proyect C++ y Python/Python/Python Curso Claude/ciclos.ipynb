# Ciclos
## For y While

Aqui se hablaran sobre los ciclos, sus funcionamientos y variables para Python, el como pueden funcionar, etc.

- **Ciclo For**

Este se compone por un codigo tipo `for` y tiene diferentes usos, como por ejemplo:

    - Listas
    - Condiciones

Tambien puedes implementar para hacer repetir cosas

- **Ciclos While**

Estos sirven principalmente para saber si algo es verdadero o falso utilizando la palabra ´while´ donde este repetira lo que tenga adentro hasta que sea *verdadero*, si aplica un *false* este hara un cuello de botella dentro del codigo

**Ejemplo de ciclo For**

*Como lista*

```python
for i in range(10):
    print("El numero actual es: " + str(i))
```

Tambien se pueden poner listas dentro del ciclo *for* de la siguiente forma

```python
frutas = ["manzana", "banana", "cereza", "naranja", "pera"]
for fruta in frutas:
    print("La fruta actual es: " + fruta)
```

**Ejemplos de Ciclo *While***

```python
i = 0
while i < 10:
    print("El numero actual es: " + str(i))
    i += 1
```

## Formas de *romper* o *saltar* una parte del Codigo en un ciclo

Para esta parte se utilizaran dos cosas:

- *Break*
- *Continue*

El primero funciona para *romper* el codigo de forma *directa*, donde este al poner una *variable tipo if* hace que se rompa.

Y el segundo funciona para *saltar* directamente una parte del codigo haciendo que esa parte no se mencione

**Ejemplo de *break***

```python
for i in range(20):
    print("El numero actual es: " + str(i))
    if i == 12:
        break
```

**Ejemplo de *Continue***

```python
for i in range(10):
    if i == 4:
        continue
    print("El numero actual es: " + str(i))
```

Tambien el ciclo *while* puede funcionar para esperar si algun usuario tiene que hacer algo con un numero de intentos, y si no funciona manda otro dato o a otra parte 

```python
i = 0
prompt = "Ingrese contraseña:"
while True:
    password = input(prompt)
    if i == 2:
        print("Demasiados intentos. Acceso denegado.")
        break
    if password == "secreto":
        print("Contraseña correcta. Acceso concedido.")
        break
    else:
        print("Contraseña incorrecta. Intente de nuevo.")
        i += 1
    
```

Otra cosa, dentro de los ciclos ***for*** tambien se pueden aplicar recorridos de lista, en donde podemos poner tanto un **numerador** como un **objeto**, como por ejemplo:

```python
frutas = ["pera", "manzana", "platano", "kiwi", "mandarina"]
for i, fruta in enumerate(frutas):
    print("El indice es: " + str(i) + " y la fruta es: " + fruta)
```
