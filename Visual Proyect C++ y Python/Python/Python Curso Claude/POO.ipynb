### Este es se utiliza con el modulo ***class*** el cual a travez de un ***molde*** se creara un **objeto**

```python
class CuentaBancaria:
    def __init__(self,nombre,saldo = 0,deuda = 0):
        self.deuda = deuda
        self.saldo = saldo
        self.persona = nombre
    def verCuenta(self):
        print(f"Hola, {self.persona}, espero tengas un buen dia, tu saldo es de {self.saldo} y tu deuda es de {self.deuda}")
    def pagarDeuda(self, deuda, saldo):
        if deuda > 0:
            print(f"Su deuda es de {self.deuda} , desea pagarlo?")
            pregunta = input(prompt = "Desea pagar su deuda?")
            if pregunta == "si" or "Si" or "SI":
                pagoDeuda = int(input(prompt = "Cuanto desea pagar?"))
                if pagoDeuda > self.saldo:
                    print("No puedes pagar mas de lo que debes")
                else:
                    self.deuda -= pagoDeuda
                    self.saldo -= pagoDeuda
                    print(f"Has pagado {pagoDeuda}, tu nueva deuda es de {self.deuda} y tu nuevo saldo es de {self.saldo}")
            else:
                print("No has pagado tu deuda")
        else: 
            print("No tienes deuda")
    def depositar(self, saldo):
        deposito = int(input(prompt = "Cuanto desea depositar?"))
        self.saldo += deposito
    def retirarEfectivo(self,saldo):
        retiro = int(input(prompt = "Cuando desea retirar?"))
        if self.saldo < 0:
            print("Esto le generara mas deuda")
            self.deuda += retiro
            self.saldo = 0
        elif self.deuda > 1000:
            print("No puede generar mas deuda, se cancelara la accion")
            while True:
                return()
        else:
            self.saldo -= retiro

cuenta1 = CuentaBancaria("Juan", 1000, 0)
cuenta2 = CuentaBancaria("Maria", 500, 200)
cuenta3 = CuentaBancaria("Pedro", 200, 100)

accionRequerida = int(input(prompt = "Que desea hacer el dia de hoy? 1. Crear cuenta 2. Entrar a cuenta"))

if accionRequerida == 1:
    nuevaCuenta = input(prompt = "Desea crear una nueva cuenta?")
    if nuevaCuenta == "Si" or "si":
        nombre = input(prompt = "Cual es su nombre??")
        dinero = input(prompt = "Cuanto quiere ingresar??")
        nuevaCuenta = CuentaBancaria(nombre, dinero, 0)
        print(f"Cuenta creada exitosamente, tu nombre es {nuevaCuenta.persona}, tu saldo es de {nuevaCuenta.saldo} y tu deuda es de {nuevaCuenta.deuda}")
    else:        print("No se ha creado una nueva cuenta")
elif accionRequerida == 2:
    nombreCuenta = input(prompt = "Cual es el nombre de la cuenta a la que desea entrar?")
    if nombreCuenta == cuenta1.persona:
        print(f"Bienvenido {cuenta1.persona}")
        accionCuenta = input(prompt = "Que desea hacer? 1. Ver cuenta 2. Pagar deuda 3. Depositar 4. Retirar efectivo")
        if accionCuenta == "Ver cuenta" or "ver cuenta":
            cuenta1.verCuenta()
        elif accionCuenta == "Pagar deuda" or "pagar deuda":
            cuenta1.pagarDeuda(cuenta1.deuda, cuenta1.saldo)
        elif accionCuenta == "Depositar" or "depositar":
            cuenta1.depositar(cuenta1.saldo)
        elif accionCuenta == "Retirar efectivo" or "retirar efectivo":
            cuenta1.retirarEfectivo(cuenta1.saldo)
    elif nombreCuenta == cuenta2.persona:
        print(f"Bienvenido {cuenta2.persona}")
        accionCuenta = input(prompt = "Que desea hacer? 1. Ver cuenta 2. Pagar deuda 3. Depositar 4. Retirar efectivo")
        if accionCuenta == "Ver cuenta" or "ver cuenta":
            cuenta2.verCuenta()
        elif accionCuenta == "Pagar deuda" or "pagar deuda":
            cuenta2.pagarDeuda(cuenta2.deuda, cuenta2.saldo)
        elif accionCuenta == "Depositar" or "depositar":
            cuenta2.depositar(cuenta2.saldo)
        elif accionCuenta == "Retirar efectivo" or "retirar efectivo":
            cuenta2.retirarEfectivo(cuenta2.saldo)
    elif nombreCuenta == cuenta3.persona:
        print(f"Bienvenido {cuenta3.persona}")
        accionCuenta = input(prompt = "Que desea hacer? 1. Ver cuenta 2. Pagar deuda 3. Depositar 4. Retirar efectivo")
        if accionCuenta == "Ver cuenta" or "ver cuenta":
            cuenta3.verCuenta()
        elif accionCuenta == "Pagar deuda" or "pagar deuda":
            cuenta3.pagarDeuda(cuenta3.deuda, cuenta3.saldo)
        elif accionCuenta == "Depositar" or "depositar":
            cuenta3.depositar(cuenta3.saldo)
        elif accionCuenta == "Retirar efectivo" or "retirar efectivo":
            cuenta3.retirarEfectivo(cuenta3.saldo)
```
